```python
with open("D:\\Protein-Sequence-Analysis\\data\\CA1.FASTA.txt", "r") as file:
    lines = file.readlines()

header = lines[0].strip()

sequence = ""

for line in lines[1:]:
    sequence = sequence + line.strip()

print("Header:")
print(header)
print()

print("Protein sequence:")
print(sequence)
print()

print("Protein length:", len(sequence))
```

```python
# count of amino acid 

amino_acids = {}

for amino_acid in sequence:
    if amino_acid in amino_acids:
        amino_acids[amino_acid] = amino_acids[amino_acid] + 1
    else:
        amino_acids[amino_acid] = 1

print(amino_acids)
```

```python
# count check of protein length and amino_acid count

print("Protein length:", len(sequence))
print("Total amino acid counts:", sum(amino_acids.values()))
```

```python
import pandas as pd 
import matplotlib.pyplot as plt
!pip install biopython
from Bio.SeqUtils.ProtParam import ProteinAnalysis
```

```python
df = pd.DataFrame(
    list(amino_acids.items()),
    columns =["amino_acid" , "count"]
)
print(df)
```

```python
# percentage count 

df["percentage"] = (df["count"] / len(sequence)) * 100 
print(df)
```

```python
# round of percentage 

df["percentage"] = df["percentage"].round(2)
print(df)
```

```python
# sortnig amino acid  by descing order 

sorted_df = df.sort_values(by="count" , ascending = False)     # ascending False itsd lagest to lowest 
print(sorted_df)
```

```python
# only to get most abundant

most_abundant = df.loc[df["count"].idxmax()]
print("Most abundant amino acid" , most_abundant)
```

```python
# for lowest abundant

least_abundant = df.loc[df["count"].idxmin()]
print("Least abundant amino acid" , least_abundant)
```

```python
# to get top 5 amino acid

# Top_5 =sorted_df.head(5)
# print("top 5 amino acid " , Top_5)

Top_5 = sorted_df.head(5)
print("top 5 amino acid" , Top_5)
```

```python
# Biological Interpretation
# We will analyze three simple properties:
# 1. Hydrophobic amino acids
# 2. Polar amino acids
# 3. Charged amino acids


# Amino acid groups

hydrophobic = "AVILMFW"
polar = "STNQ"
positive = "KRH"
negative = "DE"

print("Hydrophobic:", hydrophobic)
print("Polar:", polar)
print("Positively charged:", positive)
print("Negatively charged:", negative)
```

```python
hydrophobic_count = sum(sequence.count(aa) for aa in hydrophobic)
polar_count = sum(sequence.count(aa) for aa in polar)
positive_count = sum(sequence.count(aa) for aa in positive)
negative_count = sum(sequence.count(aa) for aa in negative)

print("Hydrophobic amino acids:", hydrophobic_count)
print("Polar amino acids:", polar_count)
print("Positively charged amino acids:", positive_count)
print("Negatively charged amino acids:", negative_count)
```

```python
hydrophobic_percentage = (hydrophobic_count / len(sequence)) * 100
polar_percentage = (polar_count / len(sequence)) * 100
positive_percentage = (positive_count / len(sequence)) * 100
negative_percentage = (negative_count / len(sequence)) * 100

print("Hydrophobic:", round(hydrophobic_percentage, 2), "%")
print("Polar:", round(polar_percentage, 2), "%")
print("Positive:", round(positive_percentage, 2), "%")
print("Negative:", round(negative_percentage, 2), "%")
```

```python
# Basic Protein Properties
# Calculate molecular weight


from Bio.SeqUtils.ProtParam import ProteinAnalysis

analysis = ProteinAnalysis(sequence)

molecular_weight = analysis.molecular_weight()

print("Protein length:", len(sequence))
print("Molecular weight:", round(molecular_weight, 2), "Da")

analysis = ProteinAnalysis(sequence)

molecular_weight = analysis.molecular_weight()

print("Protein length:", len(sequence))
print("Molecular weight:", round(molecular_weight, 2), "Da")
```

```python
pI = analysis.isoelectric_point()

print("Theoretical pI:", round(pI, 2))
```

```python
# Calculate additional protein properties 
# Isoelectric point (pI)

# Calculate isoelectric point

isoelectric_point = analysis.isoelectric_point()

print("Protein length:", len(sequence))
print("Molecular weight:", round(molecular_weight, 2), "Da")
print("Isoelectric point (pI):", round(isoelectric_point, 2))
```

```python
# Instability index

# Calculate instability index

instability = analysis.instability_index()

print("Instability index:", round(instability, 2))
```

```python
#Aliphatic index
 # Calculate Aliphatic Index manually

A = sequence.count("A")
V = sequence.count("V")
I = sequence.count("I")
L = sequence.count("L")

aliphatic_index = ((A + 2.9 * V + 3.9 * (I + L)) / len(sequence)) * 100

print("Alanine (A):", A)
print("Valine (V):", V)
print("Isoleucine (I):", I)
print("Leucine (L):", L)
print("Aliphatic index:", round(aliphatic_index, 2))
```

```python
# Calculate GRAVY score

gravy = analysis.gravy()

print("GRAVY score:", round(gravy, 3))
```

```python
print("===== Protein Properties =====")
print("Protein length:", len(sequence))
print("Molecular weight:", round(molecular_weight, 2), "Da")
print("Isoelectric point:", round(isoelectric_point, 2))
print("Instability index:", round(instability, 2))
print("Aliphatic index:", round(aliphatic_index, 2))
print("GRAVY:", round(gravy, 3))
```

```python
# Create a Protein Properties DataFrame

protein_properties = {
    "Property" : [
        "Protein Length",
        "Molecular Weight",
        "Isoelectrical point",
        "Instability Index",
        "Aliphatic Index",
        "GRAVY"
    ],
    "Value" : [
        len(sequence),
        round(molecular_weight,2),
        round(isoelectric_point,2),
        round(instability,2),
        round(aliphatic_index,2),
        round(gravy,3)
    ]
}

properties_df = pd.DataFrame(protein_properties)

print(properties_df)
```

## Biological Interpretation

Human carbonic anhydrase I (CA1) is a protein of 261 amino acids.

The amino-acid composition shows variation in the abundance of different
residues. Serine was the most abundant amino acid in the analyzed sequence,
while cysteine was among the least abundant.

The protein sequence was further analyzed using molecular weight,
isoelectric point, instability index, aliphatic index and GRAVY score.

Hydrophobic, polar and charged amino-acid groups were also analyzed to
provide a basic understanding of the physicochemical characteristics
of the protein.

These analyses demonstrate how a protein sequence can be converted into
quantitative features using Python, pandas and Biopython.
